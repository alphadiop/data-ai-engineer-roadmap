```python
%sql
CREATE SCHEMA IF NOT EXISTS nyc_taxi.audit;

DROP TABLE IF EXISTS nyc_taxi.audit.audit_load;

CREATE TABLE IF NOT EXISTS nyc_taxi.audit.audit_load
(
    periode INT,
    table_name STRING,
    taxi_type STRING,
    nb_rows BIGINT,
    status STRING,
    start_time TIMESTAMP,
    end_time TIMESTAMP,
    duration_seconds DOUBLE,
    message STRING
)
USING DELTA;
```
