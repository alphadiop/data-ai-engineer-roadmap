```python
### Analyse Gold
from pyspark.sql.functions import sum

silver_df = spark.table(
    "training.silver.orders_clean"
)

gold_df = (
    silver_df
    .groupBy("country")
    .agg(
        sum("amount").alias("revenue")
    )
)

### Trier
gold_df = gold_df.orderBy(
    gold_df.revenue.desc()
)


## Sauvegarder Gold
gold_df.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable(
        "training.gold.revenue_by_country"
    )


gold_df.show()
```

```python
df = spark.sql("""
    SELECT *
    FROM training.gold.revenue_by_country
    ORDER BY revenue DESC;
""")

df.show()
```
