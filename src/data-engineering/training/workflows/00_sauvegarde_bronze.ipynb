```python
### read CSV and sauveagarde Bronze
path = "/Volumes/training/bronze/sales_volume/order.csv"

df = (
    spark.read
         .option("header", True)
         .option("inferSchema", True)
         .csv("/Volumes/training/bronze/sales_volume/orders.csv")
)

display(df)
## attention : ['    order_id', 'customer_id', 'country', 'amount', 'order_date']

## Nettoyage rapide 
df = df.toDF(*[
    c.strip().lower().replace(" ", "")
    for c in df.columns
])


print(df.columns)

df.write \
  .format("delta") \
  .mode("overwrite") \
  .saveAsTable("training.bronze.orders_raw")
```
