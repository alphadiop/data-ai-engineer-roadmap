```python
def recreate_logs_table():

    spark.sql("DROP TABLE IF EXISTS training.monitoring.pipeline_logs")

    spark.sql("""
        CREATE TABLE training.monitoring.pipeline_logs
        (
            run_id STRING,
            run_timestamp TIMESTAMP,
            step STRING,
            status STRING,
            row_count BIGINT,
            message STRING
        )
        USING DELTA
    """)

    print("[RESET] table recréée proprement")

    recreate_logs_table()
```
