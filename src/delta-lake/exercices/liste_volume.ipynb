```python
from pyspark.sql.functions import from_unixtime, col

files = dbutils.fs.ls("/Volumes/dev/spark_db/datasets/spark_programming/data/")

df_files = (
    spark.createDataFrame(
        [(f.path, f.name, f.size, f.modificationTime) for f in files],
        ["path", "name", "size_bytes", "modification_time_ms"]
    )
    .withColumn(
        "modification_time",
        from_unixtime(col("modification_time_ms") / 1000)
    )
)

## catalog = dev
## schema = spark_db
## table = flight_time_raw
spark.sql("SELECT current_catalog()").show()
spark.sql(""" SHOW TABLES IN dev.spark_db """).show()

display(df_files)
```
