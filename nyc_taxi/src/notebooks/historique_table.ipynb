Comment voir réellement ce qu'OPTIMIZE fusionne ?

```python
df = spark.read.table("nyc_taxi.silver.silver_nyc_taxi")
(
df.filter("periode = 202601")
  .repartition(50)
  .write
  .format("delta")
  .mode("overwrite")
  .option("overwriteSchema", "true")
  .partitionBy("periode")
  .saveAsTable("nyc_taxi.silver.test_optimize")
)

display(df.limit(5))
```
