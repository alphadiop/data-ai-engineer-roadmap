```python
%sql
DROP TABLE IF EXISTS nyc_taxi.silver.silver_nyc_taxi;
DROP TABLE IF EXISTS nyc_taxi.gold.gold_fact_trips;
DROP TABLE IF EXISTS nyc_taxi.gold.gold_dim_date;
DROP TABLE IF EXISTS nyc_taxi.gold.gold_kpi_daily;
DROP TABLE IF EXISTS nyc_taxi.audit.audit_load ;
DROP TABLE IF EXISTS nyc_taxi.audit.audit_row_count;
```

```python
%sql
/*drop table if exists nyc_taxi.audit.audit_load;*/
CREATE TABLE IF NOT EXISTS nyc_taxi.audit.audit_load (
    run_id BIGINT NOT NULL,
    periode INT NOT NULL,
    table_name STRING NOT NULL,
    taxi_type STRING NOT NULL,
    status STRING NULL,
    start_time TIMESTAMP NULL,
    end_time TIMESTAMP NULL,
    duration_seconds DOUBLE NULL,
    error_step STRING NULL,
    message STRING NULL
)
USING DELTA;
```

```python
%sql
CREATE TABLE IF NOT EXISTS nyc_taxi.audit.audit_row_count (
    run_id BIGINT,
    periode INT,
    table_name STRING,
    row_count BIGINT,
    created_at TIMESTAMP
)
USING DELTA;
```
